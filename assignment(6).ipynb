```python
import numpy as np

# --- a
original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original array:", original)

# --- b.
subset = original[1:4]
print("b. Subset array:", subset)

# --- c.
subset[0] = 999
print("c. Modified subset:", subset)
print("   Original after subset modification:", original)

original = np.array([1, 2, 3, 4, 5, 6])

# --- d.
subset_copy = original[1:4].copy()
subset_copy[0] = 500
print("d. Modified copied subset:", subset_copy)
print("   Original after copy modification:", original)

# --- e.
matrix = np.arange(1, 13).reshape(3, 4)
print("\ne. 3x4 Matrix:\n", matrix)

# --- f.
print("\nf. Slicing operations:")
print("   • First row:", matrix[0, :])
print("   • Last row:", matrix[-1, :])
print("   • Second column:", matrix[:, 1])
print("   • Rows 1-2 and Columns 2-3:\n", matrix[1:3, 2:4])

# --- g.
flat_arr = matrix.flatten()
ravel_arr = matrix.ravel()
print("\ng. matrix.flatten():", flat_arr)
print("   matrix.ravel():  ", ravel_arr)

# --- h.
ravel_arr[0] = 777
print("\nh. Modified ravel()[0] to 777.")
print("   Original matrix affected?: YES\n", matrix)

# --- i.
flat_arr[1] = 888
print("\ni. Modified flatten()[1] to 888.")
print("   Original matrix affected?: NO\n", matrix)

# --- j.
print("\nj. Matrix attributes:")
print("   • shape:", matrix.shape)
print("   • ndim: ", matrix.ndim)
print("   • size: ", matrix.size)
print("   • dtype:", matrix.dtype)
```

```python
import numpy as np

y = np.array([40, 65, 30, 85])
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# a.
shape_X = X.shape
ndim_X = X.ndim

# b.
X_T = X.T

# c.
XTX = X_T @ X

# d.
XTX_inv = np.linalg.inv(XTX)

# e.
beta = XTX_inv @ X_T @ y

# g.
new_user = np.array([5, 40, 7])
pred_score = new_user @ beta


print(f"a. Shape of X: {shape_X}, Dimensions of X: {ndim_X}\n")
print("b. Transpose of X (X.T):\n", X_T, "\n")
print("c. Matrix Product (X.T @ X):\n", XTX, "\n")
print("d. Inverse of (X.T @ X):\n", XTX_inv, "\n")
print("e. Calculated Coefficients (Beta):\n", beta, "\n")
print(f"g. Predicted Assistance Score for new user: {pred_score:.2f}")
```
