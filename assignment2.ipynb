```python
# Assuming Roll Number: 2024101085
roll_number = "2024101085"


L = [int(digit) * 10 for digit in roll_number]
print("i. Original List L:", L)

L.append(90)

print("ii. After append(90):", L)

L.insert(2, 50)

print("ii. After insert(2, 50):", L)


L.remove(40)
print("iii. After remove(40):", L)

popped_val = L.pop(0)
print(f"iii. After pop(0) [removed {popped_val}]:", L)


L.sort()
print("iv. Ascending order:", L)

L.sort(reverse=True)
print("iv. Descending order:", L)

print("v. First three elements:", L[:3])
print("v. Last three elements:", L[-3:])

avg_L = sum(L) / len(L)
above_avg = [x for x in L if x > avg_L]
print(f"vi. Average = {avg_L:.2f} | Elements > Average:", above_avg)
```

```python
from collections import Counter
import random

# Set random seed using integer roll number
roll_seed = 2024101085
random.seed(roll_seed)

random_nums = [random.randint(100, 900) for _ in range(100)]


odd_count = sum(1 for x in random_nums if x % 2 != 0)
print(f"ii. Total odd numbers: {odd_count}")


even_count = sum(1 for x in random_nums if x % 2 == 0)
print(f"iii. Total even numbers: {even_count}")



def is_prime(n):
  if n < 2:
    return False
  for i in range(2, int(n**0.5) + 1):
    if n % i == 0:
      return False
  return True


primes_found = [x for x in random_nums if is_prime(x)]
print(
    f"iv. Total prime numbers: {len(primes_found)} | Prime numbers list:",
    primes_found,
)


most_common_num, frequency = Counter(random_nums).most_common(1)[0]
print(
    f"v. Most frequent number: {most_common_num} (Occurs {frequency} times)"
)
```
