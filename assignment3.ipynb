```python
import pandas as pd

data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}


df = pd.DataFrame(data)
df.to_csv('employees.csv', index=False)

import pandas as pd

# Load the dataset
df = pd.read_csv('employees.csv')

# a) Shape of the DataFrame
print("a) Shape of DataFrame:", df.shape)

# b) Summary of the DataFrame (data types & non-null counts)
print("\nb) Summary of DataFrame:")
df.info()

# c) Generate descriptive statistics
print("\nc) Descriptive Statistics:")
print(df.describe())

# d) Display first 5 rows and last 3 rows
print("\nd) First 5 rows:")
print(df.head(5))
print("\nLast 3 rows:")
print(df.tail(3))

# e) Calculate specific statistics
print("\ne) Statistics:")
print("i. Average salary:", df['Salary'].mean())
print("ii. Total bonus:", df['Bonus'].sum())
print("iii. Youngest employee's age:", df['Age'].min())
print("iv. Highest performance rating:", df['Rating'].max())

# f) Sort the DataFrame by Salary in descending order
df_sorted = df.sort_values(by='Salary', ascending=False)
print("\nf) Sorted by Salary (Descending):\n", df_sorted)

# g) Categorize employees based on performance rating
def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

df['Performance_Category'] = df['Rating'].apply(categorize_rating)
print("\ng) DataFrame with Performance Category:\n", df[['Name', 'Rating', 'Performance_Category']])

# h) Identify missing values in the DataFrame
print("\nh) Missing values count per column:\n", df.isnull().sum())

# i) Rename the Employee_ID column to ID
df.rename(columns={'Employee_ID': 'ID'}, inplace=True)
print("\ni) Renamed Columns:\n", df.columns.tolist())

# j) Filter employees
print("\nj) Filters:")
print("i. Experience > 5 years:\n", df[df['Years_of_Experience'] > 5][['ID', 'Name', 'Years_of_Experience']])
print("\nii. Belong to IT department:\n", df[df['Department'] == 'IT'][['ID', 'Name', 'Department']])

# k) Modify dataset by adding Tax column (10% of Salary)
df['Tax'] = df['Salary'] * 0.10
print("\nk) DataFrame with Tax column added:\n", df[['Name', 'Salary', 'Tax']])

# l) Save the modified DataFrame to a new CSV file
df.to_csv('modified_employees.csv', index=False)
print("\nl) Modified DataFrame saved to 'modified_employees.csv'.")
```
