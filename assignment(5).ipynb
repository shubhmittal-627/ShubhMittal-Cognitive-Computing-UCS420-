```python
#(a)
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Reversed:", reversed_arr)
#(b)
# i. Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]

print("x -> Most Frequent Value:", most_freq_x)
print("x -> Indices:", indices_x)

# ii. Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
max_count_y = np.max(counts_y)
most_freq_y = vals_y[counts_y == max_count_y]
indices_y = {val: np.where(y == val)[0] for val in most_freq_y}

print("y -> Most Frequent Value(s):", most_freq_y)
for val, idx in indices_y.items():
  print(f"y -> Indices for value {val}:", idx)
```

```python
import numpy as np

student_name = np.linspace(10, 100, 25)


print("Dimensions (ndim):", student_name.ndim)  # Output: 1
print("Shape:", student_name.shape)  # Output: (25,)
print("Total Elements (size):", student_name.size)  # Output: 25
print("Data Type (dtype):", student_name.dtype)  # Output: float64
print(
    "Total Bytes Consumed (nbytes):", student_name.nbytes
)


transpose_reshape = student_name.reshape(25, 1)
print("\nTransposed via reshape(25, 1):\n", transpose_reshape)
```

```python
import numpy as np

ucs420_student_name = np.array(
    [[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]]
)


mean_val = np.mean(ucs420_student_name)
median_val = np.median(ucs420_student_name)
max_val = np.max(ucs420_student_name)
min_val = np.min(ucs420_student_name)
unique_elems = np.unique(ucs420_student_name)

print("Mean:", mean_val)
print("Median:", median_val)
print("Max:", max_val)
print("Min:", min_val)
print("Unique Elements:", unique_elems)


reshaped_ucs420_student_name = ucs420_student_name.reshape(4, 3)
print("\nReshaped (4x3):\n", reshaped_ucs420_student_name)


resized_ucs420_student_name = np.resize(ucs420_student_name, (2, 3))
print("\nResized (2x3):\n", resized_ucs420_student_name)
```
